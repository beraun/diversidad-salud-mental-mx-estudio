```python
from pathlib import Path
import csv
root = Path.cwd()
if not (root/'datos').exists(): root=root.parent
with (root/'datos'/'h1_resumen_ponderado.csv').open(encoding='utf-8') as f:
    h1=list(csv.DictReader(f))
[(r['grupo'],r['prevalencia']) for r in h1]
```

## H1
Comparación de prevalencias ponderadas; diferencia 36.19 puntos porcentuales, sin causalidad.

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.barh([x['grupo'] for x in h1],[float(x['prevalencia']) for x in h1])
ax.set_xlabel('Ideación alguna vez en la vida (%)')
ax.invert_yaxis()
plt.show()
```

## H2
Los intervalos y las pruebas proceden del análisis de diseño de encuesta; aquí sólo se leen agregados.

```python
with (root/'datos'/'h2_resumen_ponderado.csv').open(encoding='utf-8') as f:
    h2=list(csv.DictReader(f))
plt.bar([int(x['tipos']) for x in h2],[float(x['prevalencia']) for x in h2])
plt.xlabel('Tipos distintos de victimización')
plt.ylabel('Ideación (%)')
plt.show()
```
